```python
# Dataset 1 — Numbers
# Q1
# Dataset
# [5, 10, 15, 20, 25, 30]
# Problem:
# Create an RDD from the dataset and multiply every number by 2.
# Expected Output
# [10, 20, 30, 40, 50, 60]



data=[5,10,15,20,25,30]
rdd=spark.sparkContext.parallelize(data)
rdd2=rdd.map(lambda x:x*2)
print(rdd2.collect())
```

```python
Sentences=[
 "spark is fast",
 "spark is powerful",
 "hadoop is distributed",
 "spark and hadoop are popular",
 "kafka is streaming"
]
```

```python
# Q18
# Problem:
# Convert all sentences into individual words.
# Expected Output
# spark
# is
# fast
# spark
# is
# powerful
# hadoop
# is
# distributed
# spark
# and
# hadoop
# are
# popular
# kafka
# is
# streaming

rdd=spark.sparkContext.parallelize(Sentences)
rdd1=rdd.flatMap(lambda x:x.split(' '))
print(rdd1.collect())
```

```python
# Q19
# Problem:
# Find the total number of words.
# Expected Output
# 17

rdd=spark.sparkContext.parallelize(Sentences)
rdd1=rdd.flatMap(lambda x:x.split(' '))
rdd2=rdd1.count()
print(rdd2)
```

```python
# Q20
# Problem:
# Find all words having more than 4 characters.
# Expected Output
# spark
# powerful
# hadoop
# distributed
# spark
# hadoop
# popular
# kafka
# streaming
rdd=spark.sparkContext.parallelize(Sentences)
rdd1=rdd.flatMap(lambda x:x.split())
rdd2=rdd1.filter(lambda x:len(x)>4)
print(rdd2.collect())
```

```python
# Q21
# Problem:
# Convert every word into uppercase.
# Expected Output
# SPARK
# IS
# FAST
# SPARK
# IS
# POWERFUL
# HADOOP
# IS
# DISTRIBUTED
# SPARK
# AND
# HADOOP
# ARE
# POPULAR
# KAFKA
# IS
# STREAMING
rdd=spark.sparkContext.parallelize(Sentences)
rdd1=rdd.map(lambda x:x.upper())
print(rdd1.collect())
```

```python
# Q22
# Problem:
# Find the number of unique words.
# Expected Output
# 11
rdd=spark.sparkContext.parallelize(Sentences)
rdd1=rdd.flatMap(lambda x:x.split(' '))
rdd2=rdd1.distinct()
print(rdd2.count())
```

```python
Word_Count=[
 "spark is fast",
 "spark is powerful",
 "spark is easy",
 "hadoop is distributed",
 "spark is fast"
]
```

```python
# Q23
# Problem:
# Calculate the frequency of every word.
# Expected Output
# spark -> 4
# is -> 5
# fast -> 2
# powerful -> 1
# easy -> 1
# hadoop -> 1
# distributed -> 1
rdd=spark.sparkContext.parallelize(Word_Count)
rdd1=rdd.flatMap(lambda x:x.split())
rdd2=rdd1.groupBy(lambda x:x)
rdd3=rdd2.mapValues(lambda x:len(list(x)))
print(rdd3.collect())
```

```python
# Q24
# Problem:
# Find the most frequently occurring word.
# Expected Output
# is -> 5
rdd=spark.sparkContext.parallelize(Word_Count)
rdd1=rdd.flatMap(lambda x:x.split())
rdd2=rdd1.groupBy(lambda x:x)
rdd3=rdd2.mapValues(lambda x:len(list(x)))
rdd4=rdd3.max(key=lambda x:x[1])
print(rdd4)
```

```python
# Q25
# Problem:
# Find all words that occur more than once.
# Expected Output
# spark -> 4
# is -> 5
# fast -> 2

rdd=spark.sparkContext.parallelize(Word_Count)
rdd1=rdd.flatMap(lambda x:x.split())
rdd2=rdd1.groupBy(lambda x:x)
rdd3=rdd2.mapValues(lambda x:len(list(x)))
rdd4=rdd3.filter(lambda x:x[1]>1)
print(rdd4.collect())
```

```python
# Q26
# Problem:
# Calculate the total number of words.
# Expected Output
# 15

rdd=spark.sparkContext.parallelize(Word_Count)
rdd1=rdd.flatMap(lambda x:x.split())
rdd2=rdd1.count()
print(rdd2)
```

```python
Department_Salaries=[
 ("IT", 80000),
 ("HR", 60000),
 ("IT", 90000),
 ("Finance", 75000),
 ("HR", 65000),
 ("Finance", 85000),
 ("IT", 95000),
 ("Sales", 70000),
 ("Sales", 72000),
 ("IT", 88000)
]
```

```python
# Q27
# Problem:
# Group all salaries by department.
# Expected Output
# IT -> [80000, 90000, 95000, 88000]
# HR -> [60000, 65000]
# Finance -> [75000, 85000]
# Sales -> [70000, 72000]
rdd=spark.sparkContext.parallelize(Department_Salaries)
rdd1=rdd.groupBy(lambda x:x[0])
rdd2=rdd1.mapValues(lambda x2:[x2[1] for x2 in x2])
print(rdd2.collect())
```

```python
# Q28
# Problem:
# Calculate the total salary for every department.
# Expected Output
# IT -> 353000
# HR -> 125000
# Finance -> 160000
# Sales -> 142000

rdd=spark.sparkContext.parallelize(Department_Salaries)
rdd1=rdd.groupBy(lambda x:x[0])
rdd2=rdd1.mapValues(lambda x2:[x[1] for x in x2])
rdd3=rdd2.map(lambda x:(x[0],sum(x[1])))
print(rdd3.collect())
```

```python
# Q29
# Problem:
# Calculate the average salary for every department.
# Expected Output
# IT -> 88250
# HR -> 62500
# Finance -> 80000
# Sales -> 71000

rdd=spark.sparkContext.parallelize(Department_Salaries)
rdd1=rdd.groupBy(lambda x:x[0])
rdd2=rdd1.mapValues(lambda x2:[x[1] for x in x2])
rdd3=rdd2.map(lambda x:(x[0],sum(x[1])/len(x[1])))
print(rdd3.collect())
```

```python
# Q30
# Problem:
# Find the highest salary in every department.
# Expected Output
# IT -> 95000
# HR -> 65000
# Finance -> 85000
# Sales -> 72000

rdd=spark.sparkContext.parallelize(Department_Salaries)
# rdd1=rdd.groupBy(lambda x:x[0])
# rdd2=rdd1.mapValues(lambda x2:[x[1] for x in x2])
rdd1=rdd.reduceByKey(lambda x,y:max(x,y))
print(rdd1.collect())
```

```python
# Q31
# Problem:
# Find the department having the highest total salary.
# Expected Output
# IT -> 353000
rdd=spark.sparkContext.parallelize(Department_Salaries)
rdd1=rdd.groupBy(lambda x:x[0])
rdd2=rdd1.mapValues(lambda x2: [x[1] for x in x2])
rdd3=rdd2.mapValues(lambda x:sum(x))
rdd4=rdd3.max(key=lambda x:x[1])
print(rdd4)
```

```python
# Q33
# Problem:
# Calculate total revenue for every product.
# Expected Output
# Laptop -> 350000
# Mobile -> 200000
# Tablet -> 75000

rdd=spark.sparkContext.parallelize(Sales)
rdd1=rdd.map(lambda x:(x[0],x[1]*x[2]))
rdd2=rdd1.groupBy(lambda x:x[0])
rdd3=rdd2.mapValues(lambda x:sum(x2[1] for x2 in x ))
print(rdd3.collect())
```

```python
# Q35
# Problem:
# Find the product with the highest total revenue.
# Expected Output
# Laptop -> 350000

rdd=spark.sparkContext.parallelize(Sales)
rdd1=rdd.map(lambda x:(x[0],x[1]*x[2]))
rdd2=rdd1.groupBy(lambda x:x[0])
rdd3=rdd2.mapValues(lambda x:sum(x2[1] for x2 in x))
rdd4=rdd3.max(key=lambda x:x[1])
print(rdd4)
```

```python
# Q36
# Problem:
# Find products whose total revenue is greater than 100000.
# Expected Output
# Laptop -> 350000
# Mobile -> 200000

rdd=spark.sparkContext.parallelize(Sales)
rdd1=rdd.map(lambda x:(x[0],x[1]*x[2]))
rdd2=rdd1.groupBy(lambda x:x[0])
rdd3=rdd2.mapValues(lambda x:sum(x2[1] for x2 in x))
rdd4=rdd3.filter(lambda x:x[1]>100000)
print(rdd4.collect())
```

Q50
Problem:
Perform the following complete analysis:
Calculate revenue for every order.
Keep only orders with revenue greater than 50000.
Calculate total revenue for each product.
Find the product with the highest revenue.
Save the final result into a single output partition.
Expected Output
Orders retained:
O1 -> Laptop -> 100000
O2 -> Mobile -> 100000
O6 -> Laptop -> 200000
Product Revenue:
Laptop -> 300000
Mobile -> 100000
Highest Revenue Product:
Laptop -> 300000
Output:
1 part file

```python
rdd=spark.sparkContext.parallelize(Mixed_Business_Data)
# print("revenue for every order")
rdd1= rdd.map(lambda x:(x[0],x[1],x[2]*x[3]))
# for x in rdd1.collect():
#   print(x)
# print("\n")
print(" orders with revenue greater than 50000")
print("\n")

rdd2=rdd1.filter(lambda x:x[2]>50000)
for x in rdd2.collect():
  print(x)
print("\n")
print("product revenue\n")
rdd3=rdd2.map(lambda x:(x[1],x[2]))
rdd4=rdd3.reduceByKey(lambda x,y:x+y)
for x in rdd4.collect():
  print(x)
print("\n")

print("highest revenue product:")
print("\n")
rdd5=rdd4.max(key=lambda x:x[1])
print(rdd5)
print("\n")

rdd6=spark.sparkContext.parallelize([rdd5],1)
# print("output: ",rdd6.collect())
print("\n")
print("partitions:",rdd6.getNumPartitions())

rdd7=rdd6.saveAsTextFile("final")
print("file saved successfully")


```
